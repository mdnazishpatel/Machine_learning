```python
import torch
x = torch.arange(1.,10.)
x
```

```python
##Adding an extra dimension
x_reshaped = x.reshape(1,9)
x_reshaped, x_reshaped.shape
```

```python
##shape this does the same work but updates in memory if z changes then x also changes
z= x.view(1,9)
z, z.shape
```

```python
z[:,0] =5
z,x
```

```python
## stacks tensors on eachother
x_stack = torch.stack([x,x,x], dim=1)
x_stack
```

```python
## squeeze  / it's removes extra dimension
print (x_reshaped)
print(x_reshaped.shape)
y = x_reshaped.squeeze()
print(y)
print(y.shape)
```

```python
## unsqueeze / adds a single dimension
print(y)
print(y.shape)
x_unsqueeze = y.unsqueeze(dim=0)
print(x_unsqueeze)
print(x_unsqueeze.shape)
```

```python
### permute -rearrange the dimension of a target tesnor in specified order

x_original = torch.rand(size=(224,224,3)) #[height,width,color]
x_permuted = x_original.permute(2,0,1) # [color, height,width]

print(x_original.shape)
print(x_permuted.shape)
```

### Reproducility (trying to take random out of random)

```python
random_tensor_a = torch.rand(3,4)
random_tensor_b = torch.rand(3,4)
print(random_tensor_a)
print(random_tensor_b)
print(random_tensor_a == random_tensor_b)
```

```python
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
random_tensor_c = torch.rand(3,4)
torch.manual_seed(RANDOM_SEED)
random_tensor_d = torch.rand(3,4)
print(random_tensor_c)
print(random_tensor_d)
print(random_tensor_c == random_tensor_d)
```
